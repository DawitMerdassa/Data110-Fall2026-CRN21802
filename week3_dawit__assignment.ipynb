```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
plt.style.use('ggplot')
```

```python
study_hours = [1, 2, 2.5, 3, 4, 5, 5.5, 6, 7, 8]
exam_scores = [55, 60, 63, 66, 72, 78, 80, 84, 88, 94]
```

```python
crime_types = ["Aggravated Assault", "Burglary", "Larceny", "Motor Vehicle Theft", "Murder", "Rape", "Robbery"]

crime_rates = [5.7, 38.0, 66.5, 264.1, 250.7, 318.7, 1347.2]

sorted_data = sorted(zip(crime_rates, crime_types))

rates = [x[0] for x in sorted_data]
crimes = [x[1] for x in sorted_data]

plt.figure(figsize=(10, 6))

plt.barh(crimes, rates, color="steelblue")

plt.title("Crime Rates in 2023")
plt.xlabel("Offenses per 100,000 People")
plt.ylabel("Crime Type")

plt.grid(axis="x", linestyle="-", alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
crime_types = ["Aggravated Assault", "Burglary", "Larceny", "Motor Vehicle Theft", "Murder", "Rape", "Robbery"]


percent_change = [-3.3, -8.1, -4.9, 12.0, -12.0, -9.8, -0.8]

sorted_data = sorted(zip(percent_change, crime_types))

changes = [x[0] for x in sorted_data]
crimes = [x[1] for x in sorted_data]

colors = ["red" if x < 0 else "green" for x in changes]

plt.figure(figsize=(10, 6))

plt.barh(crimes, changes, color=colors)

plt.axvline(0, color="black", linewidth=1)

plt.title("Change in Crime Rates from 2022 to 2023")
plt.xlabel("Percentage Change (%)")
plt.ylabel("Crime Type")

plt.grid(axis="x", linestyle="-", alpha=0.7)

plt.tight_layout()
plt.show()
```

I think this graph gives a better idea of what changed between the two years. It makes it easy to see which crime rates went up or down.

```python
crime_types = ["Aggravated Assault", "Burglary", "Larceny", "Motor Vehicle Theft", "Murder", "Rape", "Robbery"]

percent_change = [-3.3, -8.1, -4.9, 12.0, -12.0, -9.8, -0.8]

sorted_data = sorted(zip(percent_change, crime_types))

changes = [x[0] for x in sorted_data]
crimes = [x[1] for x in sorted_data]

colors = ["red" if x < 0 else "green" for x in changes]

plt.figure(figsize=(10, 6))

bars = plt.barh(crimes, changes, color=colors)

plt.axvline(0, color="black", linewidth=1)

for bar, value in zip(bars, changes):
    if value < 0:
        plt.text(
            value - 0.5,
            bar.get_y() + bar.get_height() / 2,
            f"{value}%",
            va="center",
            ha="right"
        )
    else:
        plt.text(
            value + 0.5,
            bar.get_y() + bar.get_height() / 2,
            f"+{value}%",
            va="center",
            ha="left"
        )

plt.title("Change in Crime Rates from 2022 to 2023")
plt.xlabel("Percentage Change (%)")
plt.ylabel("Crime Type")

plt.grid(axis="x", linestyle="-", alpha=0.5)

plt.tight_layout()
plt.show()
```

I improved the graph by adding colors and percentage labels. This makes the changes easier to see.

```python
companies = ["Company A", "Company B", "Company C"]

product1 = [20, 34, 30]
product2 = [25, 32, 34]
product3 = [30, 35, 27]

x = np.arange(len(companies))
width = 0.25

plt.figure(figsize=(9, 6))

plt.bar(x - width, product1, width, label="Product 1", color="blue")
plt.bar(x, product2, width, label="Product 2", color="orange")
plt.bar(x + width, product3, width, label="Product 3", color="green")

plt.title("Product Sales by Company")
plt.xlabel("Company")
plt.ylabel("Sales")

plt.xticks(x, companies)
plt.legend()

plt.grid(axis="y", linestyle="-", alpha=0.5)

plt.tight_layout()
plt.show()
```
